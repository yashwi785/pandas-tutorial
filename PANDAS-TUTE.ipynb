```python
#CONCATENATE & MERGE DF
```

```python
df1= pd.DataFrame({'ID' :[1,2,3], 'NAME': ['A','B','C',]})
df1
```

```python
df2=pd.DataFrame({'ID': [1,2,3,4], 'SCORE':[88,98,60,100]})
df2
```

```python
pd.concat([df1,df2], axis= 0) #vertical/row level/top on top
```

```python
pd.concat([df1,df2], axis= 1) #horizontal/column level
```

```python
#merge=join
pd.merge(df1,df2, how="inner", on="ID")
```
